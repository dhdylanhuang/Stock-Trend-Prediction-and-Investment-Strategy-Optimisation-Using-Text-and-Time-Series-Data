# Unifying Emotion

```python
import pandas as pd
```

```python
tweets = pd.read_parquet("../data/dataset/stock_tweets_sentiment_emotion_nomerge.parquet")
tweets
```

```python
def unify_emotion_features(df: pd.DataFrame) -> pd.DataFrame:

    processed_df = df.copy()


    positive_emotions = ['emotion_joy']
    negative_emotions = ['emotion_anger', 'emotion_disgust', 'emotion_sadness']
    uncertainty_emotions = ['emotion_fear', 'emotion_surprize']

    processed_df['positive_emotion'] = processed_df[positive_emotions].sum(axis=1)
    processed_df['negative_emotion'] = processed_df[negative_emotions].sum(axis=1)
    processed_df['uncertainty_emotion'] = processed_df[uncertainty_emotions].sum(axis=1)

    # Percentile-based unified features if pct columns exist
    pct_cols = {c + '_pct' for c in (positive_emotions + negative_emotions + uncertainty_emotions)}
    if pct_cols.issubset(processed_df.columns):
        processed_df['positive_emotion_pct'] = processed_df[[c + '_pct' for c in positive_emotions]].sum(axis=1)
        processed_df['negative_emotion_pct'] = processed_df[[c + '_pct' for c in negative_emotions]].sum(axis=1)
        processed_df['uncertainty_emotion_pct'] = processed_df[[c + '_pct' for c in uncertainty_emotions]].sum(axis=1)

    return processed_df
```

```python

tweets_processed = unify_emotion_features(tweets)


print("DataFrame with Unified Emotion Features:")
print(tweets_processed.head())


print("\nNew DataFrame Columns:")
print(tweets_processed.columns)
```
