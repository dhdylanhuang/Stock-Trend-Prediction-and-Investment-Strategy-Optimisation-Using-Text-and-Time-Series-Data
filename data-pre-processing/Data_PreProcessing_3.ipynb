```python
import pandas as pd
import numpy as np
import re
```

```python
tweet_df = pd.read_parquet("../data/dataset/stock_tweets.parquet")
tweet_df
pd.set_option("display.max_colwidth", None)
tweet_df
```

```python
for i in range(len(tweet_df["text"])):
    ticker = re.escape(tweet_df["ticker"][i])
    more_patterns = re.compile(
        rf"(?:\s*)(?:\$)?\b{ticker}\b(?:\s*)|(?:\s*)\bATUSER\b(?:\s*)|(?:\s*)\bAT_USER\b(?:\s*)|(?:\s*)\bURL\b(?:\s*)|(?:\s*)\brt\b(?:\s*)|(?:\s*)\binc\b(?:\s*)|(?:\s*)\bie\b(?:\s*)",
        flags=re.IGNORECASE
    )
    text = more_patterns.sub(" ", tweet_df.loc[i, "text"])
    tweet_df.loc[i, "text"] = re.sub(r"\s+", " ", text).strip()
```

```python
pd.set_option("display.max_colwidth", None)
tweet_df
```

```python
tweet_df.to_parquet("../data/dataset/stock_tweet_nomerge.parquet", index=False)
```
